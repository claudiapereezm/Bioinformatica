#### 7.- Con la secuencia y la lista siguiente, escribe un algoritmo que cuente el número de veces que aparece un elemento de la lista en la secuencia. Pista: hacer un diccionario para almacenar los resultados puede ser buena idea

```python
dna = "AATGATGAACGAC" 
dinucleotides = ['AA','AT','AG','AC', 
                 'TA','TT','TG','TC', 
                 'GA','GT','GG','GC', 
                 'CA','CT','CG','CT']
```

```python
# Crear un diccionario vacío
conteo = {}

# Contar ocurrencias de cada dinucleótido en la secuencia
for dinuc in dinucleotides:
    conteo[dinuc] = dna.count(dinuc)

print(f"El número de veces que aparece cada elemento es {conteo}")
```

##### ¿Cuántos counts tiene TC?

```python
f"Tc tiene {dna.count("TC")} counts"
```

##### ¿Puedes imprimir los dinuleótidos que tengan counts igual a 2?

```python
# Diccionario de conteos
conteo = {dinuc: dna.count(dinuc) for dinuc in dinucleotides}

# Lista de dinucleótidos con count == 2
dinucs_2 = [dinuc for dinuc, count in conteo.items() if count == 2]
print(f"Los dinucleótidos que tienen 2 counts son {dinucs_2}")
```

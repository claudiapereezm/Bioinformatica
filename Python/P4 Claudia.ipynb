##### 2.- Crea dos archivos. Uno almacena los accession que empiezen por 'a' y el otro el resto

```python
accs = ['ab56', 'bh84', 'hv76', 'ay93', 'ap97', 'bd72']
# dos listas: una para los que empiezan por a y otra para el resto
a_accs = [acc for acc in accs if acc.startswith('a')]
otros_accs = [acc for acc in accs if not acc.startswith('a')]

# cada lista en un archivo
with open("accs_a.txt", "w") as f_a:
    for acc in a_accs:
        f_a.write(acc + "\n")

with open("accs_otros.txt", "w") as f_otros:
    for acc in otros_accs:
        f_otros.write(acc + "\n")

print("Archivo 'accs_a.txt' creado con:", a_accs)
print("Archivo 'accs_otros.txt' creado con:", otros_accs)
```

##### 4.- Escribe una función que cuente el porcentaje de A de una secuencia

```python
def porcentaje_A(secuencia):
    secuencia = secuencia.upper()  # por si hay minúsculas
    total = len(secuencia)
    if total == 0:
        return 0  # evita división entre cero
    count_A = secuencia.count("A")
    porcentaje = (count_A / total) * 100
    return porcentaje
```

```python
dna = "AATGATGAACGAC"
print("El porcentaje de A es:", porcentaje_A(dna))
```

##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
def porcentaje_A(secuencia):
    secuencia = secuencia.upper()        # convierte todo a mayúsculas
    total = len(secuencia)
    if total == 0:
        return 0
    count_A = secuencia.count("A")
    porcentaje = (count_A / total) * 100
    return round(porcentaje, 2)          # redondea a 2 decimales
```

```python
dna = "aAtGAtgAaCgac"
print("El porcentaje de A:", porcentaje_A(dna), "%")
```

##### ¿Y que devuelva el número de decimales que yo quiero?

```python
def porcentaje_A(secuencia, decimales=2):
    secuencia = secuencia.upper()
    total = len(secuencia)
    if total == 0:
        return 0
    count_A = secuencia.count("A")
    porcentaje = (count_A / total) * 100
    return round(porcentaje, decimales)

dna = "aAtGAtgAaCgac"

print(porcentaje_A(dna))        # por defecto 2 decimales 
print(porcentaje_A(dna, 3))     # 3 decimales 
print(porcentaje_A(dna, 0))     # sin decimales
```

##### ¿Qué pasa si no ponemos un argumento? ¿Sabrías hacer una función encapsulada?

```python
#si el argumento no tiene valor por defecto, Python lo marca como un TypeError
def suma(a, b):
    return a + b

suma(5)   # TypeError: missing 1 required positional argument: 'b'
```

```python
#si el argumento tiene valor por defecto, se usa ese valor.
def suma(a, b=0):
    return a + b

suma(5)        # 5  (usa b=0)
```

```python
suma(5, 3)     # 8
```

```python
#se puede definir una función dentro de otra y encapsular parámetros para crear funciones a medida (un closure)
def porcentaje_base_factory(base="A", decimales=2):
    base = base.upper()

    def porcentaje_base(secuencia):
        s = secuencia.upper()
        total = len(s)
        if total == 0:
            return 0
        pct = 100 * s.count(base) / total
        return round(pct, decimales)

    return porcentaje_base
```

```python
pct_A_2 = porcentaje_base_factory("A", decimales=2)
pct_G_3 = porcentaje_base_factory("G", decimales=3)

dna = "aAtGAtgAaCgac"
print(pct_A_2(dna))  
```

```python
print(pct_G_3(dna))  
```

##### Testea tu función original con assert

```python
def porcentaje_A(secuencia, decimales=2):
    secuencia = secuencia.upper()
    total = len(secuencia)
    if total == 0:
        return 0
    count_A = secuencia.count("A")
    porcentaje = (count_A / total) * 100
    return round(porcentaje, decimales)

# pruebas con assert 
assert porcentaje_A("AATGATGAACGAC") == 46.15   # ejemplo del ADN anterior
assert porcentaje_A("aaaa", 1) == 100.0         # 4 de 4 son A --> 100%
assert porcentaje_A("ttggcc") == 0.0            # 0 A --> 0%
assert porcentaje_A("") == 0                    # cadena vacía --> 0
assert porcentaje_A("aAtGAtgAaCgac", 2) == 46.15  # comprobación del caso actual

print("Todas las pruebas pasaron correctamente.")
```

#### 5.- De la lista de accessions que se proporciona obtén lo siguiente:
1. Los que tengan el número 5
2. Los que tengan la letra 'd' o la 'e'
3. Los que tengan las letras 'd' y 'e' en ese orden
4. Los que tengan las letras 'd' y 'e' en ese orden con una letra cualquiera en medio
5. Los que tengan ambas letras 'd' y 'e' en cualquier orden
6. Los que empiezan por 'x' o por 'y'
7. Los que empiezan por 'x' o 'y' y terminan por 'e'
8. Los que tienen tres o más dígitos seguidos
9. Los que terminan con 'd' seguidos de 'a' o 'r' o 'p'

```python
import re
```

```python
accs = ['xkn59438', 'yhdck2', 'eihd39d9', 'chdsye847', 'hedle3455', 'xjhd53e', '45da', 'de37dp']
```

```python
# 1. Tienen el número 5
print([a for a in accs if re.search(r'5', a)])
```

```python
# 2. Tienen la letra d o e
print([a for a in accs if re.search(r'[de]', a)])
```

```python
# 3. Tienen d y e en ese orden 
print([a for a in accs if re.search(r'd.*e', a)])
```

```python
# 4. Tienen d y e en ese orden con una sola letra en medio
print([a for a in accs if re.search(r'd.e', a)])
```

```python
# 5. Tienen ambas letras d y e en cualquier orden
print([a for a in accs if re.search(r'd', a) and re.search(r'e', a)])
```

```python
# 6. Empiezan por x o y
print([a for a in accs if re.search(r'^[xy]', a)])
```

```python
# 7. Empiezan por x o y, y terminan por e
print([a for a in accs if re.search(r'^[xy].*e$', a)])
```

```python
# 8. Tienen tres o más dígitos seguidos
print([a for a in accs if re.search(r'\d{3,}', a)])
```

```python
# 9. Terminan por d seguido de 'a' o 'r' o 'p' (da/dr/dp)
print([a for a in accs if re.search(r'd[arp]$', a)])
```
